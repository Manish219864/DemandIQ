### Exploratory Data Analysis (EDA) on Retail Datasets


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from IPython.display import display
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load all datasets
train_df = pd.read_csv('data/train.csv')
test_df = pd.read_csv('data/test.csv')
stores_df = pd.read_csv('data/stores.csv')
oil_df = pd.read_csv('data/oil.csv')
holidays_df = pd.read_csv('data/holidays_events.csv')
transactions_df = pd.read_csv('data/transactions.csv')
submission_df = pd.read_csv('data/sample_submission.csv')

# Group all datasets in a dictionary to easily run functions across all of them
datasets = {
    'Train': train_df,
    'Test': test_df,
    'Stores': stores_df,
    'Oil': oil_df,
    'Holidays': holidays_df,
    'Transactions': transactions_df,
    'Sample Submission': submission_df
}
```

```python
# 1. Preview head() for all datasets together
for name, data in datasets.items():
    print(f"\n{'='*25} {name} (Head) {'='*25}")
    display(data.head())
```

```python
# 2. Shapes of all datasets
for name, data in datasets.items():
    print(f"{name:20} -> Shape: {data.shape}")
```

```python
# 3. Missing values for all datasets
for name, data in datasets.items():
    print(f"\n--- {name} Missing Values ---")
    print(data.isna().sum())
```

```python
# 4. Duplicate counts for all datasets
for name, data in datasets.items():
    print(f"{name:20} -> Duplicates: {data.duplicated().sum()}")
```

```python
# 5. Dataset info / data types
for name, data in datasets.items():
    print(f"\n{'='*25} {name} Info {'='*25}")
    data.info()
```

```python
# 6. Unique values count
for name, data in datasets.items():
    print(f"\n--- {name} Unique Values ---")
    print(data.nunique())
```

```python
# 7. Statistical summaries for all datasets
for name, data in datasets.items():
    print(f"\n{'='*25} {name} Describe {'='*25}")
    display(data.describe())
```

```python
# 8. Combine all related CSVs into one unified DataFrame 'df'
df = train_df.merge(stores_df, on='store_nbr', how='left') \
             .merge(oil_df, on='date', how='left') \
             .merge(holidays_df, on='date', how='left') \
             .merge(transactions_df, on=['date', 'store_nbr'], how='left')

print('Unified DataFrame Shape:', df.shape)
df.head()
```
